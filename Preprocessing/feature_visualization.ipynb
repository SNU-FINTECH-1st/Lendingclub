## Boxplot 

* 파라미터 설명 : (내가 저장할 경로 , my_dataframe:넣을 데이터프레임 , 내가 저장하고 싶은 사진의 파일이름)

```python
import pandas as pd
import seaborn as sns
import matplotlib.pyplot as plt
import math
import os

def boxplot(savepoint, data, title):
    # 디렉토리 생성
    if not os.path.exists(savepoint):
        os.makedirs(savepoint)
    
    # 데이터프레임 복사
    data = data.copy()

    # 모든 열을 순회하며 문자열 데이터를 수치 데이터로 변환
    for col in data.columns:
        if data[col].dtype == 'object':
            # 문자열 데이터 중 백분율(%)을 포함한 데이터를 처리
            if data[col].str.contains('%').any():
                data[col] = data[col].str.rstrip('%').astype(float) / 100.0
            else:
                try:
                    data[col] = data[col].astype(float)
                except ValueError:
                    # 수치로 변환할 수 없는 문자열 데이터는 건너뜀
                    data = data.drop(columns=[col])

    # 박스플롯 생성
    plt.figure(figsize=(25, 20))
    plt.suptitle(title)

    for i in range(len(data.columns)):
        plt.subplot(3, math.ceil(len(data.columns) / 3), i + 1)
        sns.boxplot(data=data.iloc[:, i])
        plt.title(data.columns[i])
        plt.xticks(rotation=45)  # x축 레이블 회전

    plt.tight_layout(rect=[0, 0.03, 1, 0.95])
    plt.savefig(f"{savepoint}//{title}.png")
    plt.close()


```

```python

# 박스플롯 생성
boxplot('/Users/parkjunhui/Desktop/StaticDS/project_lendingclub', my_dataframe , 'Boxplot_visual')
```

## Kde Plot

```python
import pandas as pd
import seaborn as sns
import matplotlib.pyplot as plt
import math
import os

def kdeplot(savepoint, data, title):
    # 디렉토리 생성
    if not os.path.exists(savepoint):
        os.makedirs(savepoint)

    # 데이터프레임 복사
    data = data.copy()

    # 범주형 데이터 제거
    for col in data.columns:
        if data[col].dtype == 'object':
            # 문자열 데이터 중 백분율(%)을 포함한 데이터를 처리
            if data[col].str.contains('%').any():
                data[col] = data[col].str.rstrip('%').astype(float) / 100.0
            else:
                try:
                    data[col] = data[col].astype(float)
                except ValueError:
                    # 수치로 변환할 수 없는 문자열 데이터는 건너뜀
                    data = data.drop(columns=[col])
                    
    # KDE plot 생성
    plt.figure(figsize=(25, 10))
    plt.suptitle(title)
    for i in range(len(data.columns)):
        plt.subplot(3, math.ceil(len(data.columns) / 3), i + 1)
        sns.kdeplot(data=data.iloc[:, i], fill=True)
        plt.title(data.columns[i])
        plt.xticks(rotation=45)  # x축 레이블 회전

    plt.tight_layout(rect=[0, 0.03, 1, 0.95])
    plt.savefig(f"{savepoint}//{title}.png")
    plt.close()
```

```python


# KDE plot 생성
kdeplot('/Users/parkjunhui/Desktop/StaticDS/project_lendingclub', my_dataframe, 'kdeplot')
```

## Group_histogram

```python

def group_histogram(savepoint, data, group, title):
    # 히스토그램 생성
    plt.figure(figsize=(25, 20))
    plt.suptitle(title)

    for i in range(len(data.columns)):
        plt.subplot(3, math.ceil(len(data.columns) / 3), i + 1)
        sns.histplot(data=data, palette='Set1', x=data.columns[i], hue=group)
        plt.xticks(rotation=45)  # x축 레이블 회전
    
    plt.tight_layout(rect=[0, 0.03, 1, 0.95])
    plt.savefig(f"{savepoint}/{title}.png")
    plt.close()
```
